## Visualize OEC

### Environment Setup

```python
# Import package
import os
import sys

import numpy as np

from pathlib import Path

from matplotlib import pyplot as plt

# Path setup
root_path: Path = Path("/Users/joseph/Desktop/NTU Course/114-2/Data Assimilation/Project")
file_path: Path = root_path / "Files"
fig_path : Path = root_path / "Figure"
```

### Load simulated observation

```python
# Load data
sim_obs: np.ndarray = np.load(file_path / "simulated_observations.npz")["X"]
```

### Setup matrix

```python
X_oec: np.ndarray = np.diag(np.ones(40)*0.1)
Y_oec: np.ndarray = np.diag(np.ones(40*8)*0.05)
Z_oec: np.ndarray = np.diag(np.ones(40*8*8)*0.01)
```

### Visualization

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

X_pcm = axes[0, 0].pcolormesh(X_oec, cmap="viridis", shading="auto", vmin=0, vmax=0.1)
axes[0, 0].set_title("OEC for X", fontsize=14)
fig.colorbar(X_pcm, ax=axes[0, 0])


axes[0, 1].imshow(Y_oec, cmap="viridis")
axes[0, 1].set_title("OEC for Y", fontsize=14)
axes[1, 0].imshow(Z_oec, cmap="viridis")
axes[1, 0].set_title("OEC for Z", fontsize=14)

```
